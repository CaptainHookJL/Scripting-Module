# Literals

##### Boolean values

- Python has two Boolean literals: `True` and `False`.
- Comparisons produce Boolean values. Use `==` to compare values; `=` assigns a value to a variable.


```python
print(1 > 2)
print(1 == 2)
```

```python
print (2>1)
print ("A" == "A")
```

```python
print(True > False)
print(True < False)
```

### Escape Characters




```python
# Two ways to write I'm happy.
print('I\'m happy.')
print("I'm happy.")
```

### Summary of Learning

- Literals represent fixed values in code.
- Numeric literals include integers (e.g., `321`, `-1`) and floating-point numbers (e.g., `2.34`).
- String literals represent text (e.g., `"I am a student"`).
- Boolean literals are `True` and `False`. In numeric contexts, they behave like `1` and `0`, respectively.
- Integer literals can use binary (base 2, `0b1010`), octal (base 8, `0o12`), or hexadecimal (base 16, `0xA`) notation. All three examples represent decimal `10`.
- Escape sequences include `\'` for a single quote, `\n` for a newline, and `\t` for a tab.
- `print()` displays its arguments. Keyword arguments such as `sep` and `end` control the output format.
