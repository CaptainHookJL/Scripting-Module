##### Question 1: You are given the following variables and their corresponding values:

a =20
b =30
c =40
d =50

Take a number between 20 and 50 using input(). Check whether the number falls in the range of a and d first. Then find whether it is beween a and b, b and c or c and d.  

```python
a = 20
b = 30
c = 40
d = 50

number = int(input("Enter an integer between 20 and 50: "))
if a <= number <= d:
    # Assign shared boundaries to the lower interval: [20, 30], (30, 40], (40, 50].
    if number <= b:
        print(f"{number} is between {a} and {b} (inclusive).")
    elif number <= c:
        print(f"{number} is greater than {b} and at most {c}.")
    else:
        print(f"{number} is greater than {c} and at most {d}.")
else:
    print(f"{number} is outside the range {a} to {d}.")
```

#### Q3 Write and test a program that creates a copy of a list of values, with each vowel replaced with an integer.  Use a function for creating the list.  

- The integer, stored in a variable which you must declare, has the initial value 1. Each time it is used to replace a vowel it is incremented by 1. For example, the list list1 is [‘a’,’b’,’c’,’d’,’e’], list2 will be [1, ‘b’,’c’,’d’,2]; if list1 is [‘a’,’a’,’a’], list2 will be [1,2,3], etc.  

```python
def replace_vowels(values):
    """Create a new list, replacing successive vowels with 1, 2, 3, ... ."""
    result = []
    next_integer = 1
    for value in values:
        if isinstance(value, str) and value.lower() in ('a', 'e', 'i', 'o', 'u'):
            result.append(next_integer)
            next_integer += 1
        else:
            result.append(value)
    return result

list1 = ['a', 'b', 'c', 'd', 'e']
list2 = replace_vowels(list1)
print("Original list:", list1)
print("New list:", list2)
print("Repeated vowels:", replace_vowels(['a', 'a', 'a']))
```

#### Q4 Create a function that takes two non-default arguments: student_name, student_id, and one default argument module_name = "Scripting for Cyber".

- Call this function with your name and student id. 
- Print your name, ID and module name in the function. 
- Now pass student ID first when you call the function (hint: keyword argument).


The following example uses Jacob Lowe's name and student number.

```python
def show_student(student_name, student_id, module_name="Scripting for Cyber"):
    print("Name:", student_name)
    print("Student ID:", student_id)
    print("Module:", module_name)

# Keep the student ID as text.
name = "Jacob Lowe"
student_id = "C00249139"

print("Positional arguments:")
show_student(name, student_id)
print("Student ID first, using keyword arguments:")
show_student(student_id=student_id, student_name=name)
```
