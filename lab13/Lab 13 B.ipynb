 #### Q1 Write  and test a program that checks if two lists, which you must create yourself, contain the same elements, even though the elements may be in a different order. 
 
 - For example, all the elements in [1,2,3,4,5] are in [5,4,3,2,1] so you should print True. Otherwise, print False. 
 
- Hint: Loop through list1. If the element exists in both lists, then delete it from both lists. At the end of the loop, if either of the lists is not empty, then the lists do not contain the same elements. In your code, write comments with different lists that you have used to test your program (e.g., # list1 = [1,2,3,4,5], list2 = [5,4,3,2,1]).

```python
def same_elements(list1, list2):
    # Work on copies so the original lists remain unchanged.
    remaining1 = list1.copy()
    remaining2 = list2.copy()
    # Iterate over a separate copy so removals do not skip elements.
    for element in list1.copy():
        if element in remaining2:
            remaining1.remove(element)
            remaining2.remove(element)
    return not remaining1 and not remaining2

# Tested: list1 = [1, 2, 3, 4, 5], list2 = [5, 4, 3, 2, 1] -> True
# Tested: list1 = [1, 2, 3], list2 = [1, 2, 4] -> False
# Tested: list1 = [1, 1, 2], list2 = [2, 1, 1] -> True
# Tested: list1 = [1, 1, 2], list2 = [1, 2, 2] -> False
# Tested: list1 = [], list2 = [] -> True
# Tested: list1 = [1, 2], list2 = [1, 2, 3] -> False
list1 = [1, 2, 3, 4, 5]
list2 = [5, 4, 3, 2, 1]
print(same_elements(list1, list2))
```
