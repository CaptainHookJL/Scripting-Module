## Q1 Convert a list to a dictionary. E.g. Given [5, 6, 7], produce {5:25, 6:36, 7:49}


```python
numbers = [5, 6, 7]
squares = {number: number ** 2 for number in numbers}
print(squares)
```

### Q2 Reverse key with value in a dictionary: E.g. Given {5:25, 6:36, 7:49}, produce {25:5, 36:6, 49:7}


```python
# The values in this example are unique, so each can become a key.
reversed_squares = {value: key for key, value in squares.items()}
print(reversed_squares)
```

### Q3 You are given the following text 

text = """The first Irish inter-county ploughing contest was held between County Wexford and County Kildare
on 16 February 1931 in a 26-acre field at Coursetown in Athy in County Kildare.
Since then, the National Ploughing Championships has expanded to over 800 acres with 1,700 exhibitors.
It has been extended beyond displays relating to ploughing, farming and machinery and now includes a tented trade 
village, live entertainment and music, fashion shows, a craft village, cooking demonstrations,
sheep dog trials, pony games and other activities. In 2014, the event attracted what was then a record of
279,500 visitors, rising to 281,000 in 2015, 283,000 in 2016 and 291,500 in 2017.
The original mission statement for the National Ploughing Association, the body which organises the event,
was set out as bringing 'the message of good ploughing to all parts of the country and to provide a pleasant 
and friendly place to meet and do business' """

- 1. Create a dictionary such that keys: words and values: frequency of the words

- 2. Print those words whose frequncies are greater than 2.

- 3. Create another dictionary from the dictionary you created in (1) with those keys whose length is gretaer than 5 characters.




Word counts ignore letter case and surrounding punctuation. Internal hyphens and numeric commas are retained.

```python
import string

text = "The first Irish inter-county ploughing contest was held between County Wexford and County Kildare\non 16 February 1931 in a 26-acre field at Coursetown in Athy in County Kildare.\nSince then, the National Ploughing Championships has expanded to over 800 acres with 1,700 exhibitors.\nIt has been extended beyond displays relating to ploughing, farming and machinery and now includes a tented trade \nvillage, live entertainment and music, fashion shows, a craft village, cooking demonstrations,\nsheep dog trials, pony games and other activities. In 2014, the event attracted what was then a record of\n279,500 visitors, rising to 281,000 in 2015, 283,000 in 2016 and 291,500 in 2017.\nThe original mission statement for the National Ploughing Association, the body which organises the event,\nwas set out as bringing 'the message of good ploughing to all parts of the country and to provide a pleasant \nand friendly place to meet and do business' "

# Count case-insensitively and strip punctuation from the ends of words.
# Keep internal hyphens and commas, such as inter-county and 1,700.
words = [word.strip(string.punctuation).lower() for word in text.split()]
words = [word for word in words if word]
frequencies = {}
for word in words:
    frequencies[word] = frequencies.get(word, 0) + 1
print("Word frequencies:", frequencies)

print("Words occurring more than twice:")
for word, count in frequencies.items():
    if count > 2:
        print(word, count)

long_words = {word: count for word, count in frequencies.items() if len(word) > 5}
print("Words longer than five characters:", long_words)
```
