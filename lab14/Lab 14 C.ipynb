# Create a frozenset with all English alphabets.
# Create a dictionary where every two consecutive alphabets are keys (a,b) and their code points (ASCII) as value


Use the 26 lowercase English letters and all adjacent pairs, including overlapping pairs such as (a, b) and (b, c).

```python
import string

alphabet = frozenset(string.ascii_lowercase)
print("Alphabet frozenset:", ''.join(sorted(alphabet)))

# Include all 25 adjacent pairs: (a, b), (b, c), ... , (y, z).
letters = sorted(alphabet)
alphabet_codes = {
    (first, second): (ord(first), ord(second))
    for first, second in zip(letters, letters[1:])
}
print("Adjacent letter pairs and ASCII code points:", alphabet_codes)
```

# Create  adictionary where name and student number are considered as key and students marks in two different modules as value. 

e.g. 
key = a forzen set like {John, C00012322}  value = a tuple like (84, 45)

Student: Jacob Lowe, C00249139. The marks (84, 45) are illustrative values from the question.

```python
# The marks below are sample data, not actual student results.
student_key = frozenset({'Jacob Lowe', 'C00249139'})
student_marks = {student_key: (84, 45)}
print("Student:", sorted(student_key))
print("Sample marks for two modules:", student_marks[student_key])
```

### Q3 Given tuple that cotains several lists of numbers. Sort each list.

 Input: ([7, 5, 4], [0,2,3, -4], [7, 5, 4, 5])
 
 Output: ([4, 5, 6], [-4, 0, 2, 3], [4, 5, 5, 7])

Correction: sorting [7, 5, 4] produces [4, 5, 7], preserving the original numbers.

```python
numbers = ([7, 5, 4], [0, 2, 3, -4], [7, 5, 4, 5])
# The question's first output list contains a typo: 7 must remain 7.
sorted_lists = tuple(sorted(values) for values in numbers)
print(sorted_lists)
```
